# Decision Trees

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
```

## 1. Decision Tree Regression

```python
# TODO: Write a regression tree class that can build tree and make prediction
class RegressionTree:
    def __init__(self, max_depth=3):
        self.model = DecisionTreeRegressor(max_depth=max_depth, random_state=42)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict(self, X):
        return self.model.predict(X)
```

```python
csv_path = Path("Data/Lab4/study_score.csv")
for folder in (Path("."), Path(".."), Path("NSU_ML")):
    candidate = folder / csv_path
    if candidate.is_file():
        csv_path = candidate
        break

df = pd.read_csv(csv_path)
X = df[["StudyHours", "SleepHours", "CoffeeCups"]]
```

```python
reg_tree = RegressionTree()
reg_tree.fit(X, df["Score"])
reg_tree.predict(X.head(3))
```

## 2. Decision Classification

```python
# TODO: Write a classificatin tree class that can build tree and make prediction
class ClassificationTree:
    def __init__(self, max_depth=3):
        self.model = DecisionTreeClassifier(max_depth=max_depth, random_state=42)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict(self, X):
        return self.model.predict(X)
```

```python
clf_tree = ClassificationTree()
clf_tree.fit(X, df["Pass"])
clf_tree.predict(X.head(3))
```
